```python
import numpy as np
import matplotlib.pyplot as plt

# Parameter
dtau = 0.01
int0 = 0.1
s    = 1

# Prepare tau and intensity
n_tau        = np.int16(( 10 - 0 ) / dtau + 1)
tau          = np.arange(n_tau) * dtau + 0
intensity    = np.zeros(n_tau)
intensity[0] = int0

# Calculate intensity
for i in np.arange(n_tau-1):
    intensity[i+1] = intensity[i] - dtau * ( intensity[i] - s)

# Plot intensity
plt.plot(tau, intensity)
plt.plot(tau, intensity, 'o')

# Plot Analytic Solution
intensity_a = int0 * np.exp(-tau) + s * ( 1 - np.exp(-tau) )
plt.plot(tau, intensity_a)
```
