**第2回 (3) 輻射輸送方程式の波長依存性**

```python
# Radiatve Transfer Equation
# (3)
import numpy as np
import matplotlib.pyplot as plt
from pylab import rcParams

rcParams['figure.figsize'] = 8, 6
# Parameters
dx, dnu    = 0.01, 0.1
int0, s0, kap0, j0, nu0  = 0.1, 1., 0.1, 0.1, 5
gamma = 0.5
xout  = [0, 5, 10, 50, 100]

# Set Parameters
xnum     = int( 100 / dx) + 1
x        = dx * np.arange(xnum)
nnum     = int( 10 / dnu) + 1
nu       = dnu * np.arange(nnum)
into     = np.zeros( [ nnum, len(xout)] )
int_t    = int0 * np.ones(nnum)
step     = 0
ostep    = []
for ii in np.arange(len(xout)):
    ostep.append( abs( x - xout[ii]).argmin() )

# Absorption & Emission Coefficient
kap = kap0 * gamma**2 / ( (nu - nu0)**2 + gamma**2 )
j   = j0   * gamma**2 / ( (nu - nu0)**2 + gamma**2 )

# Calculation
for ii in np.arange(xnum): #Loop for x
    for jj in np.arange(nnum): #Loop for nu
        int_t[jj] = int_t[jj] - kap[jj] * int_t[jj] * dx + j[jj] * dx
    if ii in ostep:
        into[:,step] = int_t
        step = step + 1

# Plot
for ii in np.arange(len(xout)):
    plt.plot(nu, into[:,ii])

plt.plot([0, 10], [1, 1], linestyle='dashed', color='red')
plt.xlim([0, 10])
plt.ylim([0, 1.2])
```

```python
# Estimation of Population Distribution
# (2)
import numpy as np
import matplotlib.pyplot as plt
from pylab import rcParams

# Parameters
num = 100
rcParams['figure.figsize'] = 10, 5

# Sample Distributions
dist_rnd  = np.random.rand(num)
dist_nrnd = np.random.randn(num)

# Calcilation
ave_rnd   = np.average(dist_rnd)
ave_nrnd  = np.average(dist_nrnd)
sgm_rnd   = np.sqrt( np.var(dist_rnd, ddof=1) )
sgm_nrnd  = np.sqrt( np.var(dist_nrnd, ddof=1) )
err_rnd   = np.sqrt( np.var(dist_rnd, ddof=1) ) / np.sqrt(num)
err_nrnd  = np.sqrt( np.var(dist_nrnd, ddof=1) )  / np.sqrt(num)

# Plot Resutls
plt.subplot(1,2,1)
plt.hist(dist_rnd, bins=12, range=(-.1, 1.1))
plt.axvline(x=ave_rnd, ymin=0, ymax=num, color='red')
plt.axvline(x=ave_rnd - sgm_rnd, ymin=0, ymax=num, color='blue', linestyle='dotted')
plt.axvline(x=ave_rnd + sgm_rnd, ymin=0, ymax=num, color='blue', linestyle='dotted')
plt.axvline(x=ave_rnd - err_rnd, ymin=0, ymax=num, color='green', linestyle='dotted')
plt.axvline(x=ave_rnd + err_rnd, ymin=0, ymax=num, color='green', linestyle='dotted')
print('Random Dist.: Ave. '+ '{:.3f}'.format(ave_rnd)+', Sgm. '+'{:.3f}'.format(sgm_rnd)+', Err. '+'{:.3f}'.format(err_rnd))

plt.subplot(1,2,2)
plt.hist(dist_nrnd, bins=14, range=(-4, 4))
plt.axvline(x=ave_nrnd, ymin=0, ymax=num, color='red')
plt.axvline(x=ave_nrnd - sgm_nrnd, ymin=0, ymax=num, color='blue', linestyle='dotted')
plt.axvline(x=ave_nrnd + sgm_nrnd, ymin=0, ymax=num, color='blue', linestyle='dotted')
plt.axvline(x=ave_nrnd - err_nrnd, ymin=0, ymax=num, color='green', linestyle='dotted')
plt.axvline(x=ave_nrnd + err_nrnd, ymin=0, ymax=num, color='green', linestyle='dotted')
print('Random Dist.: Ave. '+ '{:.3f}'.format(ave_nrnd)+', Sgm. '+'{:.3f}'.format(sgm_nrnd)+', Err. '+'{:.3f}'.format(err_nrnd))
```

```python
# Estimation of Population Distribution
# (3)
import numpy as np
import matplotlib.pyplot as plt
from pylab import rcParams

# Parameters
num   = [10, 100, 1000]
nn    = len(num)
count = 0
rcParams['figure.figsize'] = 15, 15 * nn

# Main Loop for Num
for tnum in num:
  # Produce Sample Distributions
  dist_rnd  = np.random.rand(tnum)
  dist_nrnd = np.random.randn(tnum)
  print('N='+'{:.0f}'.format(tnum))

  # Calcilation
  ave_rnd   = np.average(dist_rnd)
  ave_nrnd  = np.average(dist_nrnd)
  sgm_rnd   = np.sqrt( np.var(dist_rnd, ddof=1) )
  sgm_nrnd  = np.sqrt( np.var(dist_nrnd, ddof=1) )
  err_rnd   = np.sqrt( np.var(dist_rnd, ddof=1) ) / np.sqrt(tnum)
  err_nrnd  = np.sqrt( np.var(dist_nrnd, ddof=1) )  / np.sqrt(tnum)

  # Plot Resutls
  plt.subplot(2*nn,2,2*count+1)
  plt.hist(dist_rnd, bins=12, range=(-.1, 1.1))
  plt.axvline(x=ave_rnd, ymin=0, ymax=tnum, color='red')
  plt.axvline(x=ave_rnd - sgm_rnd, ymin=0, ymax=tnum, color='blue', linestyle='dotted')
  plt.axvline(x=ave_rnd + sgm_rnd, ymin=0, ymax=tnum, color='blue', linestyle='dotted')
  plt.axvline(x=ave_rnd - err_rnd, ymin=0, ymax=tnum, color='green', linestyle='dotted')
  plt.axvline(x=ave_rnd + err_rnd, ymin=0, ymax=tnum, color='green', linestyle='dotted')
  print(' Random Dist.: Ave. '+ '{:.3f}'.format(ave_rnd)+', Sgm. '+'{:.3f}'.format(sgm_rnd)+', Err. '+'{:.3f}'.format(err_rnd))

  plt.subplot(2*nn,2,2*count+2)
  plt.hist(dist_nrnd, bins=14, range=(-4, 4))
  plt.axvline(x=ave_nrnd, ymin=0, ymax=tnum, color='red')
  plt.axvline(x=ave_nrnd - sgm_nrnd, ymin=0, ymax=tnum, color='blue', linestyle='dotted')
  plt.axvline(x=ave_nrnd + sgm_nrnd, ymin=0, ymax=tnum, color='blue', linestyle='dotted')
  plt.axvline(x=ave_nrnd - err_nrnd, ymin=0, ymax=tnum, color='green', linestyle='dotted')
  plt.axvline(x=ave_nrnd + err_nrnd, ymin=0, ymax=tnum, color='green', linestyle='dotted')
  print(' Random Dist.: Ave. '+ '{:.3f}'.format(ave_nrnd)+', Sgm. '+'{:.3f}'.format(sgm_nrnd)+', Err. '+'{:.3f}'.format(err_nrnd))

  # Loop Count
  count=count+1
```
