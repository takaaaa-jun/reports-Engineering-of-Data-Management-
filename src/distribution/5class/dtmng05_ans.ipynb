```python
#パラメータ設定
import numpy as np
ave = 10.
std = 3.
num = 10
```

```python
#乱数生成
data = std*np.random.randn(num)+ave
print(data)
```

```python
#標準偏差と標準誤差の推定
est_std = 0
est_err = 0

ave_data = np.average(data)
for i in np.arange(num):
  est_std = est_std + (data[i]-ave_data)**2
  est_err = est_err + (data[i]-ave_data)**2

est_std = np.sqrt( est_std / (num-1) )
est_err = np.sqrt( est_err / (num-1) / num)

print("EST_STD: "+str(ave_data))
print("EST_STD: "+str(est_std))
print("EST_ERR: "+str(est_err))
```

```python
#パラメータ設定
import numpy as np
import matplotlib.pyplot as plt
ave = 10.
std = 3.
nmax= 500
num = np.arange(nmax-1)+2
```

```python
#標準偏差と標準誤差の推定
est_std = np.zeros(nmax-1)
est_err = np.zeros(nmax-1)
est_ave = np.zeros(nmax-1)

for i in np.arange(nmax-1):
  data       = std*np.random.randn(i+2)+ave
  est_ave[i] = np.average(data)
  for j in np.arange(num[i]): #分かりやすいようにFor文だが遅い...
    est_std[i] = est_std[i] + (data[j]-est_ave[i])**2
    est_err[i] = est_err[i] + (data[j]-est_ave[i])**2
  est_std[i] = np.sqrt( est_std[i] / (num[i]-1) )
  est_err[i] = np.sqrt( est_err[i] / (num[i]-1) / num[i])
```

```python
plt.figure(figsize=[10,5])
plt.plot(est_ave)
plt.plot([2,nmax],[10,10])
plt.xlim([2,nmax])
plt.ylim([7,13])
plt.title("N vs Average")
```

```python
plt.figure(figsize=[10,5])
plt.plot(est_std)
plt.plot([2,nmax],[3,3])
plt.xlim([2,nmax])
plt.ylim([0,6])
plt.title("N vs Est_STD")
```

```python
plt.figure(figsize=[10,5])
plt.plot(est_err)
plt.xlim([2,nmax])
plt.ylim([0,2])
plt.title("N vs Est_ERR")
```

```python
plt.figure(figsize=[10,5])
plt.plot(est_err)
plt.xlim([2,nmax])
plt.ylim([0.1,2])
plt.xscale('log')
plt.yscale('log')
plt.title("N vs Est_ERR")
plt.plot([10,1000],[1,0.1],"--") # X**(-.5)のグラフ
```

```python
# おまけ
plt.figure(figsize=[10,5])
plt.errorbar(np.arange(nmax-1)+2, est_ave, yerr = 2*est_err, fmt='o')
plt.plot([2,nmax],[10,10])
plt.xlim([2,nmax])
plt.ylim([7,13])
plt.title("N vs Average+2*EST_Error")
```
