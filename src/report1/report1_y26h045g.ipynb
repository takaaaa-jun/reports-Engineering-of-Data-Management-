```python
# 1回目：裏　2回目：裏　3回目：表　4回目：表　5回目：裏
# (1)最尤推定による，wの推定結果の変化を図⽰せよ．
game_result = ["裏", "裏", "表", "表", "裏"]
omote_count = 0
ura_count = 0
w_list = []
for i in range(len(game_result)):
    if (game_result[i] == "表"):
        omote_count += 1
    elif (game_result[i] == "裏"):
        ura_count += 1
    w = omote_count / (i + 1)
    w_list.append(w)

plt.figure(figsize=(8, 5))

# 横軸：観測回数
x = range(1, len(w_list) + 1)

# グラフ描画
plt.plot(x, w_list, marker="o", linewidth=1, label="(1) result")

# グラフ設定
plt.xticks(x)
plt.ylim(0, 1)
plt.xlabel("Number of count")
plt.ylabel("The value of w")
plt.title("The result of w")
plt.grid(True)
plt.legend()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# w の範囲
w = np.linspace(0, 1, 500)

# 各回の尤度関数
L1 = (1 - w)
L2 = (1 - w)**2
L3 = (1 - w)**2 * w
L4 = (1 - w)**2 * w**2
L5 = (1 - w)**3 * w**2

plt.figure(figsize=(8,8))

# 尤度関数
plt.plot(w, L1, label="1st")
plt.plot(w, L2, label="2nd")
plt.plot(w, L3, label="3rd")
plt.plot(w, L4, label="4th")
plt.plot(w, L5, label="5th")

# 最尤推定値
mle = [0, 0, 1/3, 1/2, 2/5]
likelihoods = [L1, L2, L3, L4, L5]

for i in range(5):
    ymax = np.max(likelihoods[i])
    plt.scatter(mle[i], ymax, s=70, marker="o")

# グラフ設定
plt.xlabel("w")
plt.ylabel("Likelihood")
plt.title("Likelihood Functions and Maximum Likelihood Estimates")
plt.grid(True)
plt.legend()

plt.show()
```
